## Sequential Flow

- By default all the lines in a python program run in sequential order.
- This is called **Sequential Flow**.
- Hence the order in which we write the lines of code matters.

````python
print(1)
print(2)
print(3)
print(4)
````

````python
first_name = "Angel"
print(first_name)
print(second_name)
second_name = "Gupta"
````

## Conditional Flow

- But in some cases we can alter the flow of execution of the program.
- This is called **Conditional Flow**, as it is dependent on some condition.
- We have two types of conditional flows:
    - Selective Flow
    - Iterative Flow

**Selective Flow**

- In selective flow, we select if we want to run (or skip) a particular block of code based on some conditions.

`if` keyword is used to select if you want to run (or skip) a particular block of code.

```Python
# Syntax

code_before_if_block

if (condition):
    block_of_code

code_after_if_block
```

- if the condition is True, the block of code is executed.
- if the condition is False, the block of code is skipped.

````python
hungry = True

print("wake up")

if (hungry):
    print("eat something")

print("go to sleep")
````

`else` keyword is used along with `if` to provide an alternative block of code to run if the condition is false.

```Python
# Syntax

code_before_if_block

if (condition):
    first_block_of_code

else:
    second_block_of_code

code_after_if_block
```

- if the condition is True, the first block of code is executed.
- if the condition is False, the second block of code is executed.

````python

print("grab a biscuit")

if (hungry):
    print("eat it")
else:
    print("double it and give to the next person")

print("go to sleep")
````

`elif` keyword is used along with `if` to check multiple conditions in given order until one of them is true.

```Python
# Syntax
code_before_if_block

if (first_condition):
    first_block_of_code

elif (second condition):
    second_block_of_code

else:
    third_block_of_code

code_after_if_block
```

- if the first_condition is True, the first block of code is executed.
- if the first_condition is False and the second_condition is True, the second block of code is executed.
- if both all the conditions are False, the third block of code is executed.

````python
is_stupid = True
is_hungry = True

print("grab a biscuit")

if (is_hungry):
    print("eat it")
elif (is_stupid):
    print("throws it away")

else:
    print("double it and give to the next person")

print("go to sleep")
````

````python
test_tomorrow = True
special_premiere = True
festival = False
feeling_unwell = True

print("wake up")

if(special_premiere):
  print("let's watch movie")
elif(test_tomorrow):
  print("let's study")
elif(festival):
  print("let's play with friends")
elif(feeling_unwell):
  print("let's have some rest")
else:
  print("no idea what to do")

print("go to sleep")
````

**Iterative Flow**

- In iterative flow, we iterate (repeat) a particular block of code while some condition is True.

`while` keyword is used to repeat a particular block of code, as far as the condition is True.

```Python
# Syntax

code_before_while_block

while (condition):
    block_of_code

code_after_while_block
```

- while the condition is True, the block of code is executed, then it again checks the condition.
- if the condition is False, the block of code is skipped.


````python
hungry = True

print("wake up")

while (hungry):
    print("eat something")

print("go to sleep")
````

Note:
- if you don't provide a way to eventually make your condition False, then the program will run forever.
- this is called an **infinite loop**.

````python
age = 15

while (age < 18):
    print(f"{age = }, too young for the license")
    age = age + 1

print("You can have a license")
````
