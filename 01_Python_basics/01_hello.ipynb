## `print()` function

- `print` function is used to print something on the screen
- when we try to print something with quotes "_", Python prints it as it is
- when we try to print something without quotes _, Python tries to print its value

```python
# Syntax
print("123")
print(123)
print("abc")
print(abc)  # causes error as abc is not defined
```

## Variables

- `variables` are used to store values
- think of variables as we are giving names to the values

```python
y = 10          # this assigns a value to the variable x
print("y")      # prints the text as it is
print(y)        # tries to print its value
print("y =", y) # prints both in the same line
```

Calculator — version 1

- in this version, we directly write the numbers in the code
- if we have to change the numbers, we have to change the code itself, which is not very practical

```python
a = 5
b = 8
print(a, "+", b, "=", a + b)
print(a, "-", b, "=", a - b)
print(a, "*", b, "=", a * b)
print(a, "/", b, "=", a / b)
```

## `input()` function

- `input` function is used to get input at the run time
- this allows running our code for various inputs without changing the code
- this always stores the data in the text format (string)

```python
# Syntax
var = input("custom message: ")
print("var =", var)
```

Calculator — version 2

- in this version, we use take input at run time
- we can run the same code for various inputs, without a needing to change the code
- but since the input function stores the data in text format, we are unable to operate on them as numbers

```python
a = input("Enter first number: ")
b = input("Enter second number: ")
print("The value of a is", a)
print("The value of b is", b)
print(a, "+", b, "=", a + b)
```

## Type Casting

- the reason for this behaviour is that `input()` function always stores the data as text
- even if the user enters a number, it gets converted to a string
- addition of string is different from addition of numbers

```python
a = input("Enter first number: ")
print("value of a =", a)
print("type of a =", type(a))
```

- to explicitly convert the input to a number, we use the `int()` function along with the `input()` function
- this forces the input to be converted into an integer
- but if the input cannot be converted to an integer, it throws an error
- the process of converting a value from one data type to another is called `type casting`

```python
a = int(input("Enter first number: "))
print("value of a =", a)
print("type of a =", type(a))
```

Calculator — version 3

- in this version, we use take input at run time
- this time we also convert the input to int type so that it properly handles numerical addition

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("The value of a is", a)
print("The value of b is", b)
print(a, "+", b, "=", a + b)
print(a, "-", b, "=", a - b)
print(a, "*", b, "=", a * b)
print(a, "/", b, "=", a / b)
```
