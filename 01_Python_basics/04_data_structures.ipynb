## Data Structures

These are data types which supports storing multiple values in a single variable.

- `list`: A list is a collection of items in a particular order.
- It is mainly used to stored ordered data.


- `dictionary`: A dictionary is a collection of key-value pairs.
- It is mainly used to store labelled data.

````python
# List
marks = [12, 34, 56, 78, 90]

groceries = [
    "milk",
    "bread",
    "butter",
    "toothpaste"
]

mixed = [1, 2, "hello", 1.2, True]

print(marks)
print(type(marks))
````

````python
# Dictionary
groceries = {
    "butter"    : 3,
    "bread"     : 2,
    "toothpaste": 1,
    "milk"      : 5,
    "milk"      : 3,
}

print(groceries)
print(type(groceries))
````

**Indexing**: is the address of an element within a collection


-  For `lists`, we use two types of indexing:
    - 0-based indexing
    - negative indexing
- For `dictionaries`, we use
    - key-based indexing

````python
marks = [6, 4, 8, 0, 9, 8, 7, 3]

# 0-based indexing
print(marks[0])
print(marks[1])
print(marks[2])
print()
# negative indexing
print(marks[-1])
print(marks[-2])
print(marks[-3])
````

````python
groceries = {"butter": 3, "bread": 2, "toothpaste": 1, "milk": 3}

# key-based indexing
print(groceries["bread"])
print(groceries["butter"])
print(groceries["milk"])
````

Think of it this way:

- In a list, the elements are stored and an index value is assigned to it, which is an integer starting from 0.
- In a dictionary, we can store the values and also assign some custom labels to each value, by which we can refer to them later.

````python
team = [
    "Ria",
    "Ahana",
    "Shreya",
    "Ishaan"
]
print(team[0])
print(team[2])
print()

team = {
    "Captain":      "Ria",
    "WicketKeeper": "Ahana",
    "Batsman":      "Shreya",
    "Bowler":       "Ishaan"
}
print(team['Captain'])
print(team['Bowler'])
````

## Operations on Lists and Dictionaries

- `len()`: Returns the number of elements in a collection

````python
team = [
    "Ria",
    "Ahana",
    "Shreya",
    "Ishaan"
]
print(len(team))

team = {
    "Captain":      "Ria",
    "WicketKeeper": "Ahana",
    "Batsman":      "Shreya",
    "Bowler":       "Ishaan"
}
print(len(team))
````

For lists:
- `append(value)`: Adds an value to the end of the collection
- `pop()`: Removes the last value of the collection
- `pop(index)`: Removes the value of the collection at given index
- `remove(value)`: Removes the first occurrence of the value

````python
myList = [5, 2, 1, 4, 2, 1]

print(myList)
myList.append(6)
print(myList)
myList.append(7)
print(myList)

myList.pop()
print(myList)
myList.pop(0)
print(myList)

myList.remove(1)
print(myList)
myList.remove(1)
print(myList)
````

For Dictionaries:
- `myDict[key] = value`: does either of the two things
    - if the key does not exist, it creates a new key-value pair
    - if the key exists, it updates the value
- `pop(key)`: Removes the value of the collection at given key

````python
book = {'a':'apple', 'b':'ball'}

print(book)
book['c'] = 'cat'
print(book)
book['c'] = 'cow'
print(book)

book.pop('b')
print(book)
# book.pop('d') # causes error
# print(book)
````

- `in` operator is used to check if a value exists in a collection
- for lists: it checks the value in a list
- for dictionaries: it checks the key in a dictionary

````python
nums = [5, 2, 1, 4]
book = {'a': 'apple', 'b': 'ball', 'c': 'cat'}

print(4 in nums)
print(6 in nums)
print()
print('a' in book)
print('apple' in book)
print('apple' in book.values())
print('d' in book)
````

## For loops & range objects

`for` keyword is used to repeat a particular block of code for each element in a collection

```Python
# Syntax

code_before_for_block

for var in collection:
    block_of_code

code_after_for_block
```

The difference between while loops & for loops is that while loops are used to repeat a particular block of code until a certain condition is met, while for loops are used to repeat a particular block of code for each element in a collection.

- condition -> while loop
- collection -> for loop

````python
# for loop on a list
nums = [5, 2, 1, 4]

for x in nums:
    print(f"current value of x = {x}")

# the loop runs the number of times the length of the list
# in each round, the variable x holds the value of an element in the collection
````

````python
# for loop on a dictionary
book = {'a': 'apple', 'b': 'ball', 'c': 'cat'}

for x in book:
    print(f"current value of x = {x}")
    print(f"current value of book[x] = {book[x]}")
    print()
````

````python
# for loop on a string
name = "Angel"

for x in name:
    print(f"current value of x = {x}")
````

- `range()` function is used to create a range object
- A range objects is just a collection of numbers in a given range

```python
range(start, stop, step)
```

- `start` refers to the start of range (inclusive)
- `stop` refers to the end of range (exclusive)
- `step` refers to the direction of the range (+1 for forward, -1 for backward)

- if only a single number is passed, it is considered as the stop. The start defualts to 0, the step defaults to 1.
- if two numbers are passed, it is considered as start and stop. The step defaults to 1

````python
# for loop on range objects

for x in range(5):
    print(f"current value of x = {x}")
print()

for x in range(5, 10):
    print(f"current value of x = {x}")
print()

for x in range(10, 5, -1):
    print(f"current value of x = {x}")
````

## Nested Structures

- `Nested Structures` are structures that contain other structures in them.
- Following the four simplest nested structures:
    - list of lists
    - list of dictionaries
    - dictionary of lists
    - dictionary of dictionaries

````python
# list of lists
nums = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(nums)
print(nums[2])
print(nums[2][1])
````

````python
# list of dictionaries
students = [
    {'name': 'Angel','marks': 98,'class': 12},
    {'name': 'Shreya','marks': 89,'class': 11},
    {'name': 'Ahana','marks': 72,'class': 10},
]

print(students)
print(students[0])
print(students[2]['marks'])
````

````python
# dictionary of lists
students = {
    'name': ['Angel', 'Shreya', 'Ahana'],
    'marks': [98, 89, 72],
    'class': [12, 11, 10]
}

print(students)
print(students['marks'])
print(students['marks'][2])
````

````python
# dictionary of dictionaries
captains = {
    'red_house' : {'name': 'Angel','marks': 98,'class': 12},
    'grn_house' : {'name': 'Shreya','marks': 89,'class': 11},
    'blu_house' : {'name': 'Ahana','marks': 72,'class': 10},
}

print(captains)
print(captains['red_house'])
print(captains['red_house']['name'])
````
