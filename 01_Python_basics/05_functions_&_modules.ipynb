## Homework

You are given a list of numbers. Write a code using for loop to find the maximum number in the list.

```text
Example:
input  = [6, 1, 8, 0, 3]
output = 8
```

````python
# this is our given list
nums = [6, 1, 8, 0, 3]

# to start with, we consider that the first number in the list is the max number
max = nums[0]

# then we check the other numbers in the list
for x in nums:
    # if we find a number which is greater than the current max number
    if (x > max):
        # we update the max number with the newly found number
        max = x

# once the loop is over, we print the max number
print("max =", max)
````

You are given a list of numbers. Write a code using for loop to find the minimum number in the list.

```text
Example:
input  = [6, 1, 8, 0, 3]
output = 0
```

````python
nums = [6, 1, 8, 0, 3]

min = nums[0]
for x in nums:
    if (x < min):
        min = x

print("min =", min)
````

## Functions

- A **function** is control block in programming that performs a specific task.
- We define the function only once and then reuse it as many times as we want, by just calling it by its name.
- It enhances code reusability.

````python
# Example 1:

# This is how we define a function
def say_hello(name):
    print(">> Hello", name)

# This is how we call a function
print("something ...")
say_hello("Angel")
print("something ...")
say_hello("Shreya")
print("something ...")
say_hello("Ahana")
print("something ...")
````

````python
# Example 2:

# This is how we define a function
def add(nums):
    sum = 0
    for x in nums:
        sum = sum + x
    return sum

# This is how we call a function
total = add(   [1, 2, 3, 4, 5]   )
print("total =", total)

total = add(   [-6, 1, 8, 0, -3]   )
print("total =", total)
````

**Paramaters**

- Parameters are the values that are passed to a function when it is called.
- By passing a parameter to a function, we pass a value to the function to work with.
- By passing different values to a function, we can get different results from the same function.
- We can also assign some default values to the parameters.

**Return values**

- Return values are the values that are returned from a function when it is called.
- By returning a value from a function, we can save the value in a variable for later use.
- If don't explicitly return a value, then the function returns a `None` value.

````python
def noob_function(x, y):    # without return
    sum = x + y
    print(sum)

def pro_function(x, y):     # with return
    sum = x + y
    return sum

print("Using noob_function:")
result = noob_function(6, 7)
print("result =", result)
print()
print("Using pro_function:")
result = pro_function(6, 7)
print("result =", result)
````

````python
def make_tea(boiling_time, add_sugar, tea_amount=2):
    tea = f"The tea is boiled for {boiling_time} minutes"
    if (add_sugar):
        tea += " with sugar"
    else:
        tea += " without sugar"
    tea += f" and {tea_amount} spoons of tea leaves"
    return tea
````

````python
my_tea = make_tea(5, True)
print(f"My tea = {my_tea}")

your_tea = make_tea(3, False, 3)
print(f"Your tea = {your_tea}")
````

## Modules

- A **module** is a file containing Python definitions for `functions`, `classes` and `variables`.
- We can import a module to use its functions, classes and variables in other files without needing to write them again.

````python
# importing a module
import assets.my_module

# using the module
print(assets.my_module.PI)
````

- Writing the the name of the module and the name of the function/variable in the same line is called **dot notation**.
- The dot notation can be a little tiring if the module name is long. So we use **aliasing** to make it easier.
- *Aliasing* is when you give a shorter name to a module.

````python
import assets.my_module as mod

print(mod.PI)
print(mod.HCF(12, 20))
print(mod.LCM(12, 20))
````
