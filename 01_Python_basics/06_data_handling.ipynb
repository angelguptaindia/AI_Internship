## Intro to CSV & Pandas

**Tabular data**
- Tabular data consists of data stored in rows and columns
- rows are also called records, as they represent a single observation
- rows are usually identified by their index i.e. the row number
- columns are also called fields, as they represent a single attribute
- columns are usually identified by their labels i.e. the column name

**CSV format**
- CSV stands for comma-separated values.
- Its a common format for storing tabular data.
- In a csv file, we store tabular by storing:
    - each row in the table in a new line
    - each value in a row is separated by a commas

Example:
```csv
name,age,marks
angel,15,98
shreya,14,89
ahana,13,72
```

**Pandas**
- Pandas is a library/module for working with tabular data.
- It provides a special data type (called a DataFrame) for storing tabular data and supports tabular operations.
- We can directly load a csv file into a DataFrame using the `read_csv()` function of the pandas module

````python
import pandas

table = pandas.read_csv("./assets/dataset.csv")
table
````

## Tabular Operations

---
- Checking the size of the table

    ```Python
    SYNTAX:
    rows, cols = table.shape
    ```

````python
rows, cols = table.shape
print(f"Table has {rows} rows and {cols} columns")
````

---
- Accessing a particular value *(given its row number and its column name)*

    ```Python
    SYNTAX:
    table.loc[ [row_number] , [column_name] ]
    ```

````python
result = table.loc[ [3] , ['species_ID'] ]
result
````

---
- Accessing particular columns *(given column names)*
- The same syntax can also be used for rearranging or deleting the columns

    ```Python
    SYNTAX:
    table.loc[ : , [column_names] ]
    ```

````python
result = table.loc[ : , ['species_ID', 'weight', 'height'] ]
result
````

---
- Accessing a particular rows *(given row numbers)*

    ```Python
    SYNTAX:
    table.loc[ [row_numbers] , : ]
    ```

````python
result = table.loc[ [3, 4, 5] , : ]
result
````

---
- Accessing particular rows *(given a condition)*
- This process is also known as filtering the rows

    ```Python
    SYNTAX:
    table.loc[ (condition), : ]
    ```

````python
result = table.loc[ (table['HP']>100) , : ]
result
````

---
- Rearranging rows *(based on column values)*
- This process is also known as sorting the rows

    ```Python
    SYNTAX:
    table.sort_values(['column_names'], ascending=[True/False])
    ```

````python
result = table.sort_values(['HP'], ascending=[False])
result
````

---
- Aggregate operations *(column wise)*
- This only works on numeric columns

    ```Python
    SYNTAX:
    table.describe()
    ```

````python
result = table.describe()
result
````
