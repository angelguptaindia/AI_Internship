## Data Types

- In real life, data may come in various forms - sometimes its number, sometimes its text
- to handle the same in python we have various data types:

```text
int   (integer) to store whole integer values, eg: -2, -1, 0, 1, 2,
float (decimal) to store numbers with fractional parts, eg: 3.14, -2.7, 1.0,
str   (string)  to store text, eg: "Hello", "Python", "12",
bool  (boolean) to store True or False values, eg: True, False
```

````python
i = 5
print(type(5))
print(type(i))
````

````python
f = 5.7
print(type(5.7))
print(type(f))
````

````python
s = "Angel"

print(type("Angel"))
print(type(s))
````

````python
b = False

print(type(False))
print(type(b))
````

## Operators

- `operators` are special symbols that are used to perform operations on variables and values
- the input to these operations are called _operands_
- the output of these operations is called _result_
- based on the types of the operands & result, we categorize the operators as:

````python
# Arithmetic Operators
# input : Numbers
# output: Number
# [ +, -, *, / ]

print(22 + 7)
print(22 - 7)
print(22 * 7)
print(22 / 7)

print(2**3)

print((19)//10)
````

````python
# Relational Operators
# input : Numbers
# output: Bool
# [ <, >, ==, != ]

print(22 < 7)
print(22 > 7)
print(22 == 7)
print(22 != 7)
````

````python
# Logical Operators
# input : Bool
# output: Bool
# [ not, and, or ]

print(not True)
print(not False)

A = False
B = False
print(A and B)

A = True
B = True
print(A or B)
````

## Conditions

- `conditions` are just expressions that result in a boolean value
- conditions help us converge multiple data values into a single boolean (True/False) value

````python
# ṣimple conditions

print(True)             # direct True/False value

print(1 < 2)            # reducing two numerical values into a single bool value

print(True and False)   # reducing two boolean values into a single bool value
````
